```python
onehot_inputs = []
onehot_hp1 = []
onehot_hp2 = []
for chunk_start in range(start,end,128):
    onehot_inputs.append(
        dna_io.one_hot_dna_from_file(
            chromosome=chromosome,
            start=chunk_start,
            end=chunk_start + 896,
            ref_genome=ref_genome,
            cpg=True,
            cpg_bedmethyl=CpG_pileup,
        )
    )
    onehot_hp1.append(
        dna_io.one_hot_dna_from_file(
            chromosome=chromosome,
            start=chunk_start,
            end=chunk_start + 896,
            ref_genome=ref_genome,
            cpg=True,
            cpg_bedmethyl=pileup_hp1,
        )
    )
    onehot_hp2.append(
        dna_io.one_hot_dna_from_file(
            chromosome=chromosome,
            start=chunk_start,
            end=chunk_start + 896,
            ref_genome=ref_genome,
            cpg=True,
            cpg_bedmethyl=pileup_hp2,
        )
    )
```
