```python
from dimelo import plot_read_browser,parse_bam
import matplotlib.pyplot as plt
```

```python
ref_genome = '/clusterfs/nilah/oberon/jupyter/chm13.draft_v1.0.fasta'
bam_hp1_allcontext = '/clusterfs/nilah/oberon/datasets/deep_ctcf/phased/megalodon/deep_ctcf_mod_mappings_merge_HP1_chrX.retagged.sorted.bam'
bam_hp2_allcontext = '/clusterfs/nilah/oberon/datasets/deep_ctcf/phased/megalodon/deep_ctcf_mod_mappings_merge_HP2_chrX.retagged.sorted.bam'
```

```python
region_size = 64000
region1 = f'chrX:{130_048_000}-{130_048_000+region_size}'
region2 = f'chrX:{147_776_000}-{147_776_000+region_size}'
```

```python
extract_hp1,_ = parse_bam.extract(
    input_file=bam_hp1_allcontext,
    output_name='HP1_allcontext_processed',
    ref_genome=ref_genome,
    regions=[region1,region2],
    motifs=["A,0", "CG,0"],
)
extract_hp2,_ = parse_bam.extract(
    input_file=bam_hp2_allcontext,
    output_name='HP2_allcontext_processed',
    ref_genome=ref_genome,
    regions=[region1,region2],
    motifs=["A,0", "CG,0"],
)
```

```python
plot_read_browser.plot_read_browser(
    mod_file_name=extract_hp1,
    region=region1,
    motifs=["A,0", "CG,0"],
    thresh=190,
    sort_by=["read_start"],
)
```

```python
plot_read_browser.plot_read_browser(
    mod_file_name=extract_hp2,
    region=region1,
    motifs=["A,0", "CG,0"],
    thresh=190,
    sort_by=["read_start"],
)
```

```python
plot_read_browser.plot_read_browser(
    mod_file_name=extract_hp1,
    region=region2,
    motifs=["A,0", "CG,0"],
    thresh=128,
    sort_by=["read_start"],
)
```

```python
plot_read_browser.plot_read_browser(
    mod_file_name=extract_hp2,
    region=region2,
    motifs=["A,0", "CG,0"],
    thresh=128,
    sort_by=["read_start"],
)
```
